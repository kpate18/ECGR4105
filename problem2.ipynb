```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_csv('D3.csv')

# Prepare the data
X = data[['X1', 'X2', 'X3']].values
y = data['Y'].values.reshape(-1, 1) # Ensure y is a column vector
m = len(y)

# Add intercept term to X
X_b = np.c_[np.ones((m, 1)), X]

# Initialize parameters
theta = np.zeros((X_b.shape[1], 1))

# Set hyperparameters
learning_rate = 0.05 # Using a single learning rate for simplicity
num_iterations = 1000

print(f"Starting Gradient Descent with Learning Rate: {learning_rate}, Iterations: {num_iterations}")

# Gradient Descent loop
for iteration in range(num_iterations):
    # Prediction
    predictions = X_b @ theta

    # Error
    errors = predictions - y

    # Gradients
    gradients = (1/m) * X_b.T @ errors

    # Update theta
    theta = theta - learning_rate * gradients

# Final parameters
print("\nGradient Descent Completed.")
print(f"Final Linear Model Parameters (Theta):\nIntercept: {theta[0,0]:.4f}, X1: {theta[1,0]:.4f}, X2: {theta[2,0]:.4f}, X3: {theta[3,0]:.4f}")

# Make a simple prediction
new_X_simple = np.array([[1, 1, 1, 1]]) # Add intercept (1)
predicted_y_simple = new_X_simple @ theta
print(f"Predicted Y for X1=1, X2=1, X3=1: {predicted_y_simple[0,0]:.4f}")

# Calculate predictions
predictions_full = X_b @ theta

# Plot
plt.figure(figsize=(10, 7))
plt.scatter(y, predictions_full, alpha=0.6, label='Actual vs. Predicted')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2, label='Perfect Prediction Line') # Reference line

plt.title('Actual vs. Predicted Y Values')
plt.xlabel('Actual Y')
plt.ylabel('Predicted Y')
plt.grid(True)
plt.legend()
plt.show()
```
