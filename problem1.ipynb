```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# Read the CSV file
data = pd.read_csv("D3.csv", header=None)

# Get X1, X2, X3 and Y from the file
# The first row of the CSV contains headers, so we skip the first element and convert to float
x1 = data[0].values[1:].astype(float)
x2 = data[1].values[1:].astype(float)
x3 = data[2].values[1:].astype(float)
y = data[3].values[1:].astype(float)


# Number of data points
m = len(y)

# Learning rate
learning_rate = 0.05

# Number of iterations
iterations = 1000

# X1 LINEAR REGRESSION
# Start theta values at zero
theta0 = 0
theta1 = 0

cost_x1 = []

for i in range(iterations):

    # Calculate predicted Y
    y_pred = theta0 + theta1 * x1

    # Calculate error
    error = y_pred - y

    # Calculate gradient
    gradient0 = (1 / m) * np.sum(error)
    gradient1 = (1 / m) * np.sum(error * x1)

    # Update theta values
    theta0 = theta0 - learning_rate * gradient0
    theta1 = theta1 - learning_rate * gradient1

    # Calculate cost
    y_pred = theta0 + theta1 * x1
    cost = (1 / (2 * m)) * np.sum((y_pred - y) ** 2)

    # Save cost
    cost_x1.append(cost)


# X1
x1_theta0 = theta0
x1_theta1 = theta1

print("X1 Model")
print("Y =", x1_theta0, "+", x1_theta1, "X1")
print("Cost =", cost_x1[-1])


# Plot X1 regression
plt.scatter(x1, y)

plt.plot(
    x1,
    x1_theta0 + x1_theta1 * x1
)

plt.xlabel("X1")
plt.ylabel("Y")
plt.title("Linear Regression for X1")
plt.show()


# Plot X1 cost
plt.plot(cost_x1)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost for X1")
plt.show()


# X2

theta0 = 0
theta1 = 0

cost_x2 = []

for i in range(iterations):

    y_pred = theta0 + theta1 * x2

    error = y_pred - y

    gradient0 = (1 / m) * np.sum(error)
    gradient1 = (1 / m) * np.sum(error * x2)

    theta0 = theta0 - learning_rate * gradient0
    theta1 = theta1 - learning_rate * gradient1

    y_pred = theta0 + theta1 * x2

    cost = (1 / (2 * m)) * np.sum((y_pred - y) ** 2)

    cost_x2.append(cost)


x2_theta0 = theta0
x2_theta1 = theta1

print("\nX2 Model")
print("Y =", x2_theta0, "+", x2_theta1, "X2")
print("Cost =", cost_x2[-1])


# Plot X2 regression
plt.scatter(x2, y)

plt.plot(
    x2,
    x2_theta0 + x2_theta1 * x2
)

plt.xlabel("X2")
plt.ylabel("Y")
plt.title("Linear Regression for X2")
plt.show()


# Plot X2 cost
plt.plot(cost_x2)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost for X2")
plt.show()


# X3
theta0 = 0
theta1 = 0

cost_x3 = []

for i in range(iterations):

    y_pred = theta0 + theta1 * x3

    error = y_pred - y

    gradient0 = (1 / m) * np.sum(error)
    gradient1 = (1 / m) * np.sum(error * x3)

    theta0 = theta0 - learning_rate * gradient0
    theta1 = theta1 - learning_rate * gradient1

    y_pred = theta0 + theta1 * x3

    cost = (1 / (2 * m)) * np.sum((y_pred - y) ** 2)

    cost_x3.append(cost)


x3_theta0 = theta0
x3_theta1 = theta1

print("\nX3 Model")
print("Y =", x3_theta0, "+", x3_theta1, "X3")
print("Cost =", cost_x3[-1])


# Plot X3 regression
plt.scatter(x3, y)

plt.plot(
    x3,
    x3_theta0 + x3_theta1 * x3
)

plt.xlabel("X3")
plt.ylabel("Y")
plt.title("Linear Regression for X3")
plt.show()


# Plot X3 cost
plt.plot(cost_x3)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost for X3")
plt.show()



# FINAL RESULTS
print("\nFinal Costs")

print("X1 Cost =", cost_x1[-1])
print("X2 Cost =", cost_x2[-1])
print("X3 Cost =", cost_x3[-1])


# Find lowest cost
if cost_x1[-1] < cost_x2[-1] and cost_x1[-1] < cost_x3[-1]:
    print("X1 has the lowest cost")

elif cost_x2[-1] < cost_x1[-1] and cost_x2[-1] < cost_x3[-1]:
    print("X2 has the lowest cost")

else:
    print("X3 has the lowest cost")
```
